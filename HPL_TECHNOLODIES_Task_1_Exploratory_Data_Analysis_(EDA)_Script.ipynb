## Introduction to Iris Flower

The **Iris flower** is a common flower used for learning data analysis and machine learning.

The Iris dataset contains information about **three types of Iris flowers**:

- `Iris-setosa`
- `Iris-versicolor`
- `Iris-virginica`

The dataset records four main measurements of each flower: **sepal length, sepal width, petal length, and petal width**.

These measurements help us understand the differences between the three types of Iris flowers.

## Sepal and Petal

An Iris flower has two important parts called the **sepal** and **petal**.

- **Sepal:** The sepal is the outer part of the flower. It helps protect the flower.
- **Petal:** The petal is the colorful part of the flower. It helps attract insects.

## Sepal Features

The Iris dataset contains two sepal measurements:

- **Sepal Length:** The length of the sepal, measured in centimeters.
- **Sepal Width:** The width of the sepal, measured in centimeters.

These are represented by `SepalLengthCm` and `SepalWidthCm` in the dataset.

## Petal Features

The Iris dataset contains two petal measurements:

- **Petal Length:** The length of the petal, measured in centimeters.
- **Petal Width:** The width of the petal, measured in centimeters.

These are represented by `PetalLengthCm` and `PetalWidthCm` in the dataset.

## Features Used in the Dataset

The four main features used in the analysis are:

1. `SepalLengthCm`
2. `SepalWidthCm`
3. `PetalLengthCm`
4. `PetalWidthCm`

These measurements describe the physical size of each Iris flower.

## Feature Columns

The Iris dataset has four main feature columns:

- **SepalLengthCm:** It represents the length of the sepal in centimeters.
- **SepalWidthCm:** It represents the width of the sepal in centimeters.
- **PetalLengthCm:** It represents the length of the petal in centimeters.
- **PetalWidthCm:** It represents the width of the petal in centimeters.

These four columns are called **features** because they provide information about the physical characteristics of each flower.

The `Species` column is the **target/label column** because it tells us which type of Iris flower the measurements belong to.

The `Id` column is only an identifier and is not considered a feature for analysis.

#Calculate correlation matrix

```python
measurement_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

correlation_matrix = df[measurement_columns].corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

### Observation

The correlation matrix shows the relationships between the numerical measurement attributes of the Iris dataset.

- `SepalLengthCm` and `PetalLengthCm` have a **strong positive correlation of 0.87**.
- `SepalLengthCm` and `PetalWidthCm` have a **strong positive correlation of 0.82**.
- `PetalLengthCm` and `PetalWidthCm` have a **very strong positive correlation of 0.96**.
- `SepalWidthCm` and `PetalLengthCm` have a **moderate negative correlation of -0.42**.
- `SepalWidthCm` and `PetalWidthCm` have a **moderate negative correlation of -0.36**.
- `SepalLengthCm` and `SepalWidthCm` have a **very weak negative correlation of -0.11**.

The strongest relationship is between `PetalLengthCm` and `PetalWidthCm`, with a correlation of **0.96**. This indicates a very strong positive linear relationship between petal length and petal width.

Overall, the analysis shows that petal measurements are strongly related to each other and are also positively related to sepal length, while sepal width has weak to moderate negative relationships with the other measurements.

#Plot correlation heatmap

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

### Observation

The correlation heatmap shows the relationships between the numerical measurement attributes of the Iris dataset.

- `SepalLengthCm` and `PetalLengthCm` have a **strong positive correlation of 0.87**.
- `SepalLengthCm` and `PetalWidthCm` have a **strong positive correlation of 0.82**.
- `PetalLengthCm` and `PetalWidthCm` have a **very strong positive correlation of 0.96**.
- `SepalWidthCm` and `PetalLengthCm` have a **moderate negative correlation of -0.42**.
- `SepalWidthCm` and `PetalWidthCm` have a **moderate negative correlation of -0.36**.
- `SepalLengthCm` and `SepalWidthCm` have a **very weak negative correlation of -0.11**.

The diagonal values are `1.00` because each variable has a perfect correlation with itself.

Overall, the heatmap shows that the strongest relationship is between `PetalLengthCm` and `PetalWidthCm`, with a correlation of **0.96**. This indicates a very strong positive relationship between petal length and petal width.

The heatmap also shows that `SepalWidthCm` has negative relationships with the other three measurements, while the remaining measurements have strong positive relationships with each other.
